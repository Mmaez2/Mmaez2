```python
import altair as alt
from vega_datasets import data
import pandas as pd
import matplotlib.pyplot as pl
url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/bfro_reports_fall2022.csv"
df = pd.read_csv(url)
```

```python
source = data.stocks()
input_dropdown = alt.binding_select(options=df["season"].unique().tolist(), name="Season: ")
dropdown_select = alt.selection_point(
    fields=["season"],
    bind=input_dropdown
)

line = alt.Chart(df).mark_line().encode(
    x="date:T",
    y="uv_index:Q",
    color="season:N",
    opacity=alt.condition(dropdown_select, alt.value(1), alt.value(0.1))
).add_params(dropdown_select)
points = alt.Chart(df).mark_circle(size=60).encode(
    x="date:T",
    y="uv_index:Q",
    color="season:N",
    tooltip=["season:N", "date:T", "uv_index:Q"],
    opacity=alt.condition(dropdown_select, alt.value(1), alt.value(0))
)

Cutie_chart = (line + points).interactive()

Cutie_chart
```

```python
alt.Chart(df, title="windies").mark_rect().encode(
    alt.X("date(date):O").title("date").axis(format="%e", labelAngle=0),
    alt.Y("month(date):O").title("Month"),
    alt.Color("max(wind_speed)").title(None),
    tooltip=[
        alt.Tooltip("monthdate(date)", title="Date"),
        alt.Tooltip("max(wind_speed)", title="windies levels"),
    ],
).configure_view(
    step=13,
    strokeWidth=0
).configure_axis(
    domain=False
)
```
